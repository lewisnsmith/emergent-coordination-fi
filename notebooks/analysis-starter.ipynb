# flock — analysis starter
Load a run, recompute cohort metrics, and poke at decision logs.
Run from the repo root (`uv run jupyter lab` after `uv add --dev jupyterlab`).

```python
from flock.analysis import convergence, coordination
from flock.logging_.decisions import resolve_run_dir

run_dir = resolve_run_dir("latest")
run = convergence.load_run(run_dir)
run["manifest"]["run_id"], run["manifest"]["n_agents"]
```

```python
cohorts = [c["name"] for c in run["manifest"]["config"]["cohorts"]]
{c: convergence.cohort_metrics(run, c) for c in cohorts}
```

```python
# decision rationales for one agent
d = run["decisions"]
d[d.cohort == "llm"][["step", "agent_id", "action", "rationale"]].head(20)
```

```python
# herding metrics per cohort
{c: coordination.coordination_metrics(d, c) for c in cohorts}
```
